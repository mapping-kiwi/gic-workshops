```python
PATH = "PATHNAME"
gdp = "r"+PATH
arcpy.env.workspace = gdb
arc.env.overwriteOutput = True
```

```python
# Define input and output feature classes
points = os.path.join(gdb, "study points")
features = os.path.join(gdb, "roads")

buffer_output = os.path.join(gdb, "point_buffer")
selected_output = os.path.join(gdb, "selected_roads")

arcpy.analysis.Buffer(
    in_features=points,
    out_features=buffer_output,
    buffer_distance_or_field="500 Meters",
)
```

```python
# Create feature layer for selection
arcpy.management.MakeFeatureLayer(
    in_features=features,
    out_layer="roads_layer"
)
```

```python
# Select features intersecting the buffer
arcpy.management.SelectLayerByLocation(
    in_layer="roads_layer",
    overlap_type="INTERSECT",
    select_features=buffer_output,
    selection_type="NEW_SELECTION"
)
```

```python
# Check how many features were selected
count = int(arcpy.management.GetCount("roads_layer")[0])
print(f"{count} roads selected!")
```

```python
# Export selected features
arcpy.management.CopyFeatures(
    in_features="roads_layer",
    out_feature_class=selected_output
)
```
